```python
import heapq
import math

"""
./venv/bin/python /home/tobiasocula/Documents/whack/WHACK-solution/sloping_waves.py

"""

import numpy as np
import pyvista as pv

def waves(num_waves=3, standard_dist=1, num_points=200):
    peaks = np.random.uniform(0, np.pi*2, size=num_waves)
    amps = np.random.uniform(1, 2, size=num_waves)
    freqs = np.arange(1, num_waves+1)
    x = np.linspace(0, np.pi*2, num_points)
    y = sum(np.cos(amps[i]*(x + peaks[i])) for i in range(num_waves))
    return y + standard_dist + np.min(y)

y = waves()

import matplotlib.pyplot as plt

x = np.linspace(0, np.pi*2, 200)
fig, ax = plt.subplots()
ax.plot(x, y)
fig.show()
```
